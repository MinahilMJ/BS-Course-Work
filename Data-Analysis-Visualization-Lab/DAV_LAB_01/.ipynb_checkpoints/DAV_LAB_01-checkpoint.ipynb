```python
#exercise # 02
def logistic(z,a,b):
    return 1/(1+np.exp(-(a*z+b)))
```

```python
def tanh(z):
    return np.tanh(z)
```

```python
def relu(z,e):
    return np.maximum(0,e*z)
```

```python
z=np.linspace(-5,5,100)
```

```python
logistic_values=logistic(z,1,0)
tanh_values=tanh(z)
relu_values=relu(z,1)
```

```python
import matplotlib.pyplot as plt
```

```python
fig,ax=plt.subplots(figsize=(10,6))

ax.plot(z,logistic_values,linestyle='-',label='Logistic')
ax.plot(z,tanh_values,linestyle='--',label='Tanh')
ax.plot(z,relu_values,linestyle=':',label='ReLU')

ax.set_xlabel("z",fontsize=12)
ax.set_ylabel("Function Value",fontsize=12)
ax.set_title("Activation Functions",fontsize=14)
ax.legend()
ax.grid(alpha=0.3)

plt.show()
```
